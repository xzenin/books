```python
import subprocess

def check_ollama_port(port='11434'):
    try:
        output = subprocess.run(['sudo', 'lsof', '-i', '-P', '-n'],
                              capture_output=True, text=True).stdout
        if f":{port} (LISTEN)" in output:
            print(f"✅ Ollama is actively listening on port {port}")
        else:
            print(f"⚠️ Ollama not detected on port {port}")
    except Exception as e:
        print(f"❌ Error checking port: {e}")
check_ollama_port()
```

```python
import os, subprocess, time, threading

os.environ["OLLAMA_HOST"] = "0.0.0.0"
ollama_proc = subprocess.Popen(
    ["ollama", "serve"],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

time.sleep(3)
print("Ollama started")
```

```python
tunnel_proc = subprocess.Popen(
    ["cloudflared", "tunnel", "--url", "http://localhost:11434", "--http-host-header", "localhost:11434"],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

time.sleep(5)
print("Tunnel started")
```

```python
import re

output = ""
for _ in range(20):
    line = tunnel_proc.stdout.readline()
    if not line:
        break
    output += line
    print(line.strip())
    m = re.search(r"https://[a-zA-Z0-9\-]+\.trycloudflare\.com", output)
    if m:
        public_url = m.group(0)
        print("PUBLIC URL:", public_url)
        break
```
